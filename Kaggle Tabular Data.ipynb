## Before you start

### What this project is for

The point is not to win the competition. It is to go through every step of a real data science
project once, from a dataset you have never seen to a result, and then to document it so that someone
else can understand what you did and pick up where you left off. The GitHub repository you build is
also a portfolio: a link an interviewer can click to see immediately what you can do.

### How a Kaggle competition works

**Kaggle** hosts datasets and competitions. A competition page has an **Overview** (the task, and the
**metric** submissions are scored on, for example the area under the ROC curve) and a **Data** tab.
Some competitions make you join and accept their rules before you can download.

The Data tab usually holds three files:

- `train`: examples **with** the answer (the target). All of your work happens here: you split it
  into your own training, validation and test samples.
- `test`: examples **without** the answer. Beware the name: *Kaggle's test set is not the test sample
  you use in your own work.* You only run your final model on it.
- a **sample submission**: the format for your predictions on `test`.

You upload your predictions, Kaggle (which knows the answers) scores them, and you appear on the
**leaderboard**. These competitions are old, so you are not competing for anything, but most still
accept **late submissions**, and getting a score back is a good sign that everything works.

Kaggle also shows other people's public notebooks for each competition. You may look at them to learn,
but don't copy someone else's work.

**Getting the data.** Downloading from the browser is fine for this project. In general your work has
to be reproducible, so a download command kept in your notebook is better (`wget`, `curl`, or the
`kaggle` command-line tool, which needs an API key and is optional here). **Never commit the data to
GitHub**: say in your README where it comes from and how to get it.

### Choosing your dataset

- **Tabular** means rows are examples and columns are features: high-level quantities, not raw
  images, audio or video, which need far more computing.
- **Not every CSV file is a table of examples.** Open the files before you choose. Kaggle's CSV data
  comes in roughly three kinds:
  - **Simple:** one table, one row per example, one column for the target. This is what the project
    is designed for.
  - **Many tables you have to correlate:** for example customers in one file, their transactions in
    another and the products in a third, linked by ID columns. Before any machine learning you have
    to join them and summarize them into one row per example (how many transactions, their total,
    and so on). That is much more work; using only some of the files can be enough.
  - **Not really tabular, though it comes as CSV:** mouse movements captured during game play, time
    series, event logs. Each row is one measurement at one moment, not one example, and an example
    (a game session, a day, a user) is many rows of varying number. It has to be summarized per
    example (for instance a mean, a standard deviation, a count) before it is a table, and choosing
    those summaries is the hard part.
- A competition comes with a task and a metric. A plain dataset doesn't: you would have to define the
  question and the target yourself, which is more work, so agree it with me first.
- **Red flags:**
  - a target that is computed from, or defined by, another column;
  - so few rows and columns that the problem is usable but not an interesting machine-learning problem;
  - a question you could answer by looking something up or by counting.
- If a dataset is too big for your computer, cut it down.
- Not the Titanic. It is popular, but it is not meaningful, and I have seen it many, many times.
- A hard dataset is not held against you: I evaluate you against what I expect of you, not against
  the leaderboard. Several of the listed competitions have pitfalls on purpose.

## Data Cleaning and Preparation for Machine Learning

* Perform any data cleaning. Be clear what are you doing, for what feature. 
* Determine if rescaling is important for your Machine Learning model.
    * If so select strategy for each feature.
    * Apply rescaling.

Things that need thought:

- **Outliers.** You cannot define outliers as "everything beyond some quantile": on a normal
  distribution that just cuts off the tails, and the tails are data. An outlier is a value that comes
  from something you weren't accounting for, or something broken. Before removing anything, ask why
  it is there, say what you removed and why, and check what changes if you put it back. Mistakes here
  are fine, as long as you share them.
- **One-hot encoding** is needed for categories with more than two values; check the number of
  columns afterwards.
- **Locations, zip codes, IDs of places** can't just be put in as numbers. Options: a separate model
  per group, the group as a one-hot input, or an ordering that means something (for example by the
  group's median target).
- **A very rare class** can be dropped, with the reason stated, or kept and handled. Either is fine.
- After cleaning and scaling, **visualize again**, to make sure you did what you meant to.

### Before you trust your result

- **Drop identifiers** (row IDs, names). A model given an ID will memorize which ID goes with which
  answer instead of learning from the features. Keep the IDs aside: the submission file needs them.
- **Start with a baseline**: a guess, a cut on one feature, or a logistic regression. If a fancier
  model then fails, you know whether to blame the data or your code.
- **"Non-trivial" means better than chance.** A result at chance could mean the problem is hard, but
  it could also mean you made a mistake; better than chance shows you didn't.
- One working model is enough. For tabular data the strongest method is usually **boosted decision
  trees**, but you don't have to use them.
- **Be suspicious of results that are too good.** Machine learning will do the easy thing if it can. A
  score of 100% (or 98% when everyone else gets much less) usually means a feature that gives the
  answer away, a leak from the target, or a trivial rule hidden in the data. Find out which.
- **With imbalanced classes**, accuracy misleads: if 98% of examples are class 0, always answering 0 is
  98% accurate. Look at each class's performance, and try weighting each class by one over its
  frequency (`class_weight`), with and without the weights.
- With many features, models other than deep networks can do worse; start with fewer features you
  understand.
- Compute the competition's own metric yourself, as an estimate of what Kaggle will report.
- Before predicting Kaggle's `test` file, **retrain your chosen model on all of `train`**. A difference
  like 0.88 against 0.89 may just be a fluctuation: retrain a few times to see.
- For a regression target, report how close the predictions are (for example, "within one year 90% of
  the time") and whether they are biased, not only one error number.
- **Submit.** It is a small part of the grade, but uploading and getting a score is the sign that the
  whole chain works, and the first thing anyone will ask about.
