*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here
def game_win(board):
    for i in range(len(board)):
        #Row checking for the winner
        if all(board[i][j] == 1 for j in range(len(board))):
            return 1
        elif all(board[i][j] == 2 for j in range(len(board))):
            return 2
    #column checking for the winner
    for j in range(len(board)):
        if all(board[i][j] == 1 for i in range(len(board))):
            return 1
        elif all(board[i][j] == 2 for i in range(len(board))):
            return 2
        
    #checking for the main diagonal
    if all(board[i][i] == 1 for i in range(len(board))):
        return 1
    elif all(board[i][i] == 2 for i in range(len(board))):
        return 2
    
    #checking for the other diagonal
    if all(board[i][len(board) - 1 - i] == 1 for i in range(len(board))):
        return 1
    elif all(board[i][len(board) - 1 - i] == 2 for i in range(len(board))):
        return 2
    
    #checking if the game is incomplete or game is draw
    for row in board:
        for cell in row:
            if cell == 0:
                return -1
            
    return 0
        
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here
import random

def computer_move(board, computer_player, human_player):
    # 1. Try to win
    for i in range(len(board)):
        for j in range(len(board[i])):
            if board[i][j] == 0:
                board[i][j] = computer_player

                if game_win(board) == computer_player:
                    return True

                board[i][j] = 0

    # 2. Try to block the human
    for i in range(len(board)):
        for j in range(len(board[i])):
            if board[i][j] == 0:
                board[i][j] = human_player

                if game_win(board) == human_player:
                    board[i][j] = computer_player
                    return True

                board[i][j] = 0

    # 3. Otherwise choose a random empty location
    empty_locations = []

    for i in range(len(board)):
        for j in range(len(board[i])):
            if board[i][j] == 0:
                empty_locations.append((i, j))

    if len(empty_locations) > 0:
        row, col = random.choice(empty_locations)
        make_move(board, computer_player, row, col)
        return True

    return False
````

````python
# Test your solution here
def play_computer_game():
    board = create_board(3, 3)

    human_player = 1
    computer_player = 2

    while True:
        # Human turn
        draw_board(board)
        print("Your turn (X)")
        player_move(board, human_player)

        status = game_win(board)

        if status == 1:
            draw_board(board)
            print("You win!")
            break

        elif status == 0:
            draw_board(board)
            print("The game is a draw!")
            break

        # Computer turn
        print("Computer's turn...")
        computer_move(board, computer_player, human_player)

        status = game_win(board)

        if status == 2:
            draw_board(board)
            print("Computer wins!")
            break

        elif status == 0:
            draw_board(board)
            print("The game is a draw!")
            break
play_computer_game()
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
def minimax(board, depth, max_depth, maximizing, computer_player, human_player):
    status = game_win(board)

    # If computer wins
    if status == computer_player:
        return 10 - depth

    # If human wins
    if status == human_player:
        return depth - 10

    # If draw
    if status == 0:
        return 0

    # Stop searching when max depth is reached
    if depth >= max_depth:
        return 0

    # Computer's turn: maximize score
    if maximizing:
        best_score = float("-inf")

        for i in range(len(board)):
            for j in range(len(board[i])):
                if board[i][j] == 0:
                    board[i][j] = computer_player

                    score = minimax(
                        board,
                        depth + 1,
                        max_depth,
                        False,
                        computer_player,
                        human_player
                    )

                    board[i][j] = 0

                    best_score = max(best_score, score)

        return best_score

    # Human's turn: minimize score
    else:
        best_score = float("inf")

        for i in range(len(board)):
            for j in range(len(board[i])):
                if board[i][j] == 0:
                    board[i][j] = human_player

                    score = minimax(
                        board,
                        depth + 1,
                        max_depth,
                        True,
                        computer_player,
                        human_player
                    )

                    board[i][j] = 0

                    best_score = min(best_score, score)

        return best_score


def computer_move_search(board, computer_player, human_player, max_depth):
    best_score = float("-inf")
    best_move = None

    for i in range(len(board)):
        for j in range(len(board[i])):
            if board[i][j] == 0:
                board[i][j] = computer_player

                score = minimax(
                    board,
                    1,
                    max_depth,
                    False,
                    computer_player,
                    human_player
                )

                board[i][j] = 0

                if score > best_score:
                    best_score = score
                    best_move = (i, j)

    if best_move is not None:
        row, col = best_move
        make_move(board, computer_player, row, col)
        return True

    return False


def play_computer_game_search(max_depth):
    board = create_board(3, 3)

    human_player = 1
    computer_player = 2

    while True:
        draw_board(board)

        print("Your turn (X)")
        player_move(board, human_player)

        status = game_win(board)

        if status == 1:
            draw_board(board)
            print("You win!")
            break

        elif status == 0:
            draw_board(board)
            print("The game is a draw!")
            break

        print("Computer's turn...")

        computer_move_search(
            board,
            computer_player,
            human_player,
            max_depth
        )

        status = game_win(board)

        if status == 2:
            draw_board(board)
            print("Computer wins!")
            break

        elif status == 0:
            draw_board(board)
            print("The game is a draw!")
            break
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Write your solution here
def computer_vs_computer(board_size, max_depth, games=10):
    smart_wins = 0
    simple_wins = 0
    draws = 0

    for game in range(games):
        board = create_board(board_size, board_size)

        simple_player = 1
        smart_player = 2

        current_player = 1

        while True:

            # Simple computer's turn
            if current_player == simple_player:
                computer_move(
                    board,
                    simple_player,
                    smart_player
                )

            # Smart minimax computer's turn
            else:
                computer_move_search(
                    board,
                    smart_player,
                    simple_player,
                    max_depth
                )

            status = game_win(board)

            # Simple computer wins
            if status == simple_player:
                simple_wins += 1
                break

            # Smart computer wins
            elif status == smart_player:
                smart_wins += 1
                break

            # Draw
            elif status == 0:
                draws += 1
                break

            # Switch player
            if current_player == 1:
                current_player = 2
            else:
                current_player = 1

    smart_win_rate = (smart_wins / games) * 100

    print(f"Board size: {board_size} x {board_size}")
    print(f"Smart computer wins: {smart_wins}")
    print(f"Simple computer wins: {simple_wins}")
    print(f"Draws: {draws}")
    print(f"Smart computer win rate: {smart_win_rate}%")
    print()

    return smart_win_rate
````

````python
# Test your solution here
rate_3x3 = computer_vs_computer(3, 9, 10)

rate_4x4 = computer_vs_computer(4, 3, 10)

rate_5x5 = computer_vs_computer(5, 2, 10)
````
