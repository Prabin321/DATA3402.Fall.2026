In this lab, **you are allowed and encouraged to use LLMs responsibly** as learning tools.

### Appropriate uses
- Asking for **explanations** of Python concepts (lists, loops, functions, conditionals)
- Getting **hints** or alternative approaches when you are stuck

- Submitting code you cannot explain

*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here
def game_win(board):
    for i in range(len(board)):
        #Row checking for the winner
        if all(board[i][j] == 1 for j in range(len(board))):
            return 1
        elif all(board[i][j] == 2 for j in range(len(board))):
            return 2
    #column checking for the winner
    for j in range(len(board)):
        if all(board[i][j] == 1 for i in range(len(board))):
            return 1
        elif all(board[i][j] == 2 for i in range(len(board))):
            return 2
        
    #checking for the main diagonal
    if all(board[i][i] == 1 for i in range(len(board))):
        return 1
    elif all(board[i][i] == 2 for i in range(len(board))):
        return 2
    
    #checking for the other diagonal
    if all(board[i][len(board) - 1 - i] == 1 for i in range(len(board))):
        return 1
    elif all(board[i][len(board) - 1 - i] == 2 for i in range(len(board))):
        return 2
    
    #checking if the game is incomplete or game is draw
    for row in board:
        for cell in row:
            if cell == 0:
                return -1
            
    return 0
        
````
