## Object Oriented Programming

*Encapsulate* data and the operations into a new concept called an object. An object
  * has a *type* referred to as a **class**, analogous to “float” or “string”.
  * holds *data* in form of fields or **attributes** (sometimes referred to as **properties**). 
  * holds *code* in form of **methods**.

It helps to see where this sits among the abstractions you already know. We abstract numbers by putting them in variables, and we abstract code by putting it in functions. An object does both at once: it marries a piece of data to the code that works on it and puts the two behind a single abstraction barrier, so that whoever uses the object just interacts with it and lets it do its work, without needing to see what is happening to the data or the code inside. That is the idea at the heart of object oriented programming, and it is the prevalent way most software is built today.

Object oriented programming is a paradigm, a way of organizing a program, rather than a feature of any one language. Most languages now build it in to make it easier, but as long as a language gives you some way to keep data together in one place, you can do object oriented programming in it; below we'll see that Python's dictionaries are enough to fake it without any of Python's object oriented machinery. (Methods are also often called **member functions**, the name C++ uses for them.)

For example, lets say we want to keep information on students in this class for computing grades. We can create a new class of object called a “student”:

  * **Attributes**: name, id number, year, grades, …
  * **Methods**: add_grade, average_grade, …
  
In python, here's an example creating such a class:


```python
class student:
    name=str()
    id_number=int()
    year=int()
    grades=list()
    
    def add_grade(self,grade):
        self.grades.append(grade)
        
    def average_grade(self):
        return sum(self.grades)/len(self.grade)
```

Any given student is an instance of the student class. Like a numbers or a strings, we can create, use, and store instances of students as we need.

Note that `self` is how we refer to data or methods of the instance of class itself. 

```python
a_student=student()

print("The student class:", type(student))
print("The student instance:", type(a_student))
```

It seems that all instances of the class seem to share some of the same attributes. That may be useful in some circumstances, but in most cases each object instance should hold its own data. The problem is that we have created the attributes at when we define the class rather than when we create a new instance of a class. We'll do that in a bit...

There are times you genuinely want data shared across every instance rather than private to each one — for example, if you wanted the class itself to keep a running count of how many students have ever been created, that count belongs to the class, not to any one instance. That's a deliberate choice, though, made by putting the data at the class level on purpose — not something that should happen by accident because you forgot to set it up per instance, which is the trap we just fell into.

### Why OO?

Why would such a construction be helpful? 

The alternative way of storing and doing all of the book-keeping of data associated with multiple students is to create a bunch of lists for each of the attributes and make sure that the first student's information is always at index 0, second student index 1, and so on. 

For exmaple:

```python
names=list()
id_numbers=list()
years=list()
grades=list()

# Create an "instance" of a student

names.append(str())
id_numbers.append(int())
years.append(str())
grades.append(list())
```

This gets worse as a student grows. The grouping of a student's information here is implicit, carried only by the convention of matching indices, rather than stated anywhere as "these things go together as a student". With a hundred attributes you would need a hundred lists to keep aligned. And adding a new attribute later means finding every place in your code that creates a student and making sure it fills in the new list as well, or the lists fall out of sync. With a class, you add the new attribute once, to the class definition, and every student has it.

We could write functions to help make all of this look nicer, but it would be cumbersome to manage and ugly to read. By encapsulating concepts into objects, we ultimately reduce the complexity of your code by abstracting everything associated with a students into a new object type (class), and storing the data of students in instances of this new class.

Note that python dictionaries would allow us to in build structures that are similar to an object oriented class, for example:

```python
students = [ { "name": "Jane Doe", "Major": "Data Science", "Grades": list(), "GetGrades": lambda self: self["Grades"]},
             { "name": "John Doe", "Major": "Data Science", "Grades": list(), "GetGrades": lambda self: self["Grades"]}]
```

Written this way it's tedious, and not as pretty as the class syntax we're about to use, but it does work. It's also close to what Python itself does underneath: its object oriented machinery is built on the same basic ideas and data structures, dictionaries holding data and functions, with a nicer syntax put in front. In a couple of lectures we'll look underneath to see how Python does it.

But with no defined structure, every data member and operation would have to managed by convention and not enforced by the langauge. That means that a type is just a convention and a code cannot assume, for example, that every student in a list of students would have exactly the same methods.

It's also worth saying plainly why this matters beyond tidiness: with either the lists-of-attributes version or the dictionary version, the structure only exists by *convention* — nothing stops another piece of code from reaching in and adding a student without an ID, or deleting from one list and not the others, and nothing in the language would catch it. Wrapping the data in a class doesn't make that mistake physically impossible on its own, but it gives you one place to enforce the rules, and — as we'll see shortly with private data — a way to actually stop outside code from reaching in at all. None of this means you should always reach for a class, either: Python's dictionaries and lists are powerful enough that quick, small pieces of data are often better left as one, in a plain dictionary. Reach for a class when you need to enforce structure or behavior, not by default.


## Constructor / Destructor

We created an instance of student in the example above, but we didn't take care to carefully make sure that the student instance was carefully setup, so all the instances shared attributes. Let's properly build our instances.

The important OO concepts here are **constructors** and **destructors**. These are optional methods that are called when an object is created or destroyed. Since python manages memory for us, we typically don't need to implement destructors, but constructors are always a good idea. 

In python the names of build-in methods of classes typically start and end with 2 underscores. `__init__(self,...)` and `__del__(self)` are class constructor and destructor, respectively. 

For example:

**Object lifecycle reminder:** `__init__` runs when an object is created. `__del__` runs when the object is about to be destroyed (when its reference count reaches zero).


```python
class student:
    def __init__(self, name, id_number, year):
        self.name=name
        self.id_number=id_number
        self.year=year
        self.grades=list()
    
    def add_grade(self,grade):
        self.grades.append(grade)
        
    def average_grade(self):
        return sum(self.grades)/len(self.grades)
    
    def print_grades(self):
        for grade in self.grades:
            print(grade)          
```

Now when you instantiate a student you would do:

```python
a_student=student("John Doe", 111, 0)

a_student.add_grade(85)
a_student.add_grade(90)

a_student.print_grades()

print("Average:", a_student.average_grade())
```

And you can keep all of the information for all of your students in a list:

```python
students=list()

students.append(student("Jim Doe", 111, 0))
students.append(student("Jane Doe", 112, 0))

print(type(students[0]))
students[0].add_grade(100)
students[0].add_grade(99)
students[1].add_grade(89)
students[1].add_grade(100)
```

```python
for student in students:
    print("Name:", student.name)
    print("Average:", student.average_grade())
```

```python
dir(student)
```

There are lots of built-in methods for classes, some of which have default implementations that you can **overload**, others that you can optionally implement. For example, if you have objects that you want python to know how to add, you can implement `__add__(self,other)` method. More on this later.

Object oriented programming allows you to establish and maintain abstractions more effectively. Once an object has been implemented, users of the object don't need to know how the internals work and how it stores data to use it. They simply use the object's methods. It also makes it easier to change implementations. As long as the class maintains the same method names (aka interface), you can improve and evolve the data and methods as you like without affecting any application that uses your class.

### Reference Counting Example

We can use reference counting for other purposes. For example, imagine you have data stored in a file or database that you want to represent in as objects in python so you can easily manipulate and organize them, but you only want to read the data into memory
when you need it.

Here's a class that demonstrates how to do it:

```python
class FileLine:
    _file_handle = None      # shared file object
    _ref_count = 0           # number of active instances
    _filename = "my_big_csv_file.csv"         # file to open

    def __init__(self, line_number):
        self.line_number = line_number

        # increment reference counter
        FileLine._ref_count += 1
        print(f"Instance created. Active objects: {FileLine._ref_count}")

    def get_line(self):
        # lazily open the file
        if FileLine._file_handle is None:
            print("Opening file...")
            FileLine._file_handle = open(FileLine._filename, "r")

        # read specific line
        FileLine._file_handle.seek(0)
        for i, line in enumerate(FileLine._file_handle, start=1):
            if i == self.line_number:
                return line.strip()

        return None

    def __del__(self):
        FileLine._ref_count -= 1
        print(f"Instance deleted. Active objects: {FileLine._ref_count}")

        if FileLine._ref_count == 0 and FileLine._file_handle is not None:
            print("Closing file...")
            FileLine._file_handle.close()
            FileLine._file_handle = None
```

Notice the two kinds of data in this class. Most of the time each instance should keep its own data, and here `line_number` does: each `FileLine` remembers which line it stands for. But the file handle and the reference count are deliberately kept at the class level and shared by every instance. This is the case mentioned earlier, where sharing data between instances is exactly what we want.

To see why, imagine the file is enormous, far too big to read into memory, and we only ever want a small subset of its lines, perhaps selected at random, so that we can work with them as objects. We don't want to open the file over and over. If we hold 10,000 lines out of a file of a million, we don't want 10,000 file handles; we want one, opened the first time any line is actually read and shared by all of those objects. The reference count is how the class knows when the last line object is gone, so the file stays open exactly as long as something still needs it.

Two caveats. The way `get_line` finds its line is not a good implementation: every time it's called, it goes back to the start of the file and reads forward until it reaches the line it wants, which would be hopeless on a truly huge file. It stands in for a proper way of jumping to a line; the idea is what matters here. And you aren't expected to be able to write a class like this yet. It's here so you can start to picture the mechanics underneath: how data that belongs to one instance differs from data shared across all of them, and how reference counting, which we come back to below, can be put to work.

Lets make an example file to test this (using some simple shell scripting):

```python
!echo "Line 1" > my_big_csv_file.csv
!echo "Line 2" >> my_big_csv_file.csv
!echo "Line 3" >> my_big_csv_file.csv
!echo "Line 4" >> my_big_csv_file.csv
!cat my_big_csv_file.csv
```

And test:

```python
line1 = FileLine(1)
line3 = FileLine(3)

print(line1.get_line())
print(line3.get_line())

del line1
del line3
```

What we see is that file file was opened only when the data was accessed, and closed once the last reference to any line was removed from memory.

Python automatically manages memory for us. When you create an object, Python allocates memory for it. Each time a variable stores that object, Python increases a reference counter.

When a reference is removed — for example, when a variable is reassigned, goes out of scope, or is explicitly deleted — the reference count is decremented. When the reference count reaches zero, Python frees the memory.

This reference-counting system is why many objects in Python are cleaned up immediately when they are no longer used (especially in CPython), though more complex situations involving circular references are handled by an additional garbage collector.

---
**Summary:**
- Objects use memory.
- Python counts how many references point to each object.
- When the count reaches zero, the object is removed.


## Classes and Instances

Consider the following simple class:

```python
class person:
    def __init__(self, name):
        self.name = name
 
    def say_hello(self):
        print("Hello, my name is " + self.name)
```

Now lets instanciate and test 2 persons:

```python
# create objects
bob = person("Bob")
bill = person("Bill")
 
# call methods owned by virtual objects
bob.say_hello()
bill.say_hello()
```

Above, we created the two instances by calling `person(...)`, and then kept the references to those instances in python variables `bob` and `bill`. 

We can also keep references to persons in a list:

```python
people = [person("Jane"),person("John")]
Jane = people[0]
Jane.say_hello()
people[1].say_hello()
```

These objects stay alive because something refers to them. Here the list holds the references to the two new persons, and the variable `people` holds the reference to the list. If we deleted `people`, the list would go, and with it any person that nothing else refers to: the second person would be deleted, but the first would survive, because `Jane` still refers to it. If you want to keep your data, keep a reference to it.

Notice too that because all of these objects are persons, they share the same interface even though each holds different data, so we can manage them all in exactly the same way, asking any one of them to say hello. That's the beginning of an idea called **polymorphism**, which we'll come back to shortly.

## Inheritance

A powerful feature of object-oriented programming is inheritance, which allows you to build a hierarchy of classes. For example, what if we wanted to keep track of students and faculty at the University. There would be some aspects of students and faculty that would be in common, while other would be different. We can store the common atrributes and methods in a common class called "person" that both "student" and "faculty" **inherit** from. 

For example:


```python
class person:
    def __init__(self, name, id_number):
        self.name=name
        self.id_number=id_number
    
    
class student(person):
    def __init__(self, name, id_number, year):
        super(student,self).__init__(name,id_number)
        # self.person(name,id_number)
        self.year=year
        self.grades=list()
    
    def add_grade(self,grade):
        self.grades.append(grade)
        
    def average_grade(self):
        return sum(self.grades)/len(self.grades)
    
    def print_grades(self):
        for grade in self.grades:
            print(grade)

            
class faculty(person):
    def __init__(self, name, id_number):
        super(faculty,self).__init__(name,id_number)
        self.courses=list()
    
    def add_courses(self,course):
        self.grades.append(course)
        
    def print_courses(self):
        for courses in self.courses:
            print(course)
```

Inheritance helps in reducing the need to copy and paste code and to make it easier to use your code by establishing common interface and behaviors between different objects.

One thing to be careful of in the example above: a parent's constructor doesn't run on its own. `person.__init__` is what sets `name` and `id_number`, and if the child's constructor never calls it, those lines are never executed and the student simply has no `name` at all. Some languages call the parent's constructor for you; Python doesn't, so you have to call it explicitly, and usually it's the first thing the child's constructor does.

## Public and Private Methods

By convention, methods that start with two underscores (`__`) are considered to be private and are meant to be only called by the class.

The reason to do this isn't secrecy for its own sake — it's that once other code depends on a class, you want to be free to change how it works *inside* without breaking everyone who uses it, as long as the outward-facing methods keep behaving the same way. That outward-facing set of methods is the class's **interface**: what you promise callers, as opposed to how you happen to have implemented the promise this week. Marking the bookkeeping methods private means no one can call them directly, reach in, and leave the object in a state your own class wouldn't have allowed.

```python
class device: 
    def __init__(self):
        self.__update()
 
    def operate(self):
        print('operate')
 
    def __update(self):
        print('updating software')
```

Here we made the update method private by calling it `__update()`. We are safe to call this function from inside of the class itself, for example in the constructor as we did here. 

```python
a_device = device()

a_device.operate()
```

But, if we try to use that method externally, we get an error:

```python
# This will fail
a_device.__update()
```

## Public and Private Data

Usually a class needs to control the data it holds. If an external class or user changes a data member of a class in a unexpected way, then the class can fail.

The way to control the data in your classes is to make the varibles holding the data private and only allow changes to data in "setter" and "accessor" functions that you ensure safely modify the data your class holds.

Consider this example of a class to represent a car.

Two things worth noticing once you see it in code: an accessor hands back a *value*, not a way to reach back in and change the original — being told a car's name doesn't give you any way to rename it. And a setter is exactly the place to validate: if you don't check that a new name is actually a string before storing it, some other part of your code that assumes `car.name()` is always a string will eventually crash on data you let in carelessly, far away from where the bad value actually came from.

One refinement of the first point. The car's accessors can safely hand back what they hold because the name is a string and the counts are numbers, and whoever receives one of those can't change it in place. A list is different: returning a list returns a reference to the very list the object keeps inside, and the caller could then append to it or delete from it and quietly break the class's bookkeeping. If you want to be strict, an accessor for a list should return a copy of it instead.

```python
class car:
    def __init__(self,name="Unnamed",n_doors=4, max_passengers=4):
        self.__n_passengers = 0
        self.__name=name
        self.__n_doors=n_doors
        self.__max_passengers=max_passengers
        
    ## Accessors / Getters
    def name(self):
        return self.__name
    
    def n_doors(self):
        return self.__n_doors
    
    def n_passengers(self):
        return self.__n_passengers

    def max_passengers(self):
        return self.__max_passengers


    ## Setter
    def set_name(self,name):
        if isinstance(name,str):
            self.__name=name
        else:
            print ("Name must be a string.")
        
    ## Can't change number of doors on a car... so no setter for __n_doors
        
    ## We can only add and remove passengers
    def add_passenger(self,n=1):
        if isinstance(n,(int,float)):
            self.__n_passengers+=n
            if self.__n_passengers>self.__max_passengers:
                self.__n_passengers=self.__max_passengers
                print ("Car is full. ",n-self.max_passengers," passengers were left outside.")
        else:
            print ("Number of passengers must be an integer.")
        
    def remove_passenger(self,n=1):
        if isinstance(n,int):
            self.__n_passengers-=n
            if self.__n_passengers<0:
                self.__n_passengers=0            
        else:
            print("Number of passengers must be an integer.")
```

Lets test this class:

```python
my_car=car()
print (my_car.name())
my_car.set_name("Geronimo")
print (my_car.name())

my_car.add_passenger()
print (my_car.n_passengers())
```

## Method Overloading

**Overloading** refers to the ability to define new data or methods for a child class that "overload" the same data or methods of a parent class. A common coding pattern is to create a shared parent class for two or more children classes. The parent class defines a set of data and methods which are either not implemented at all (referred to as **virtual** method) or provide some default behavior. The child classes overload these methods to implement their own specializations. 

For example:

```python
class person:
    __name=""
    def __init__(self, name):
        self.__name=name
    
    # This is a virtual method
    def do_work(self):
        raise NotImplementedError
    
class student(person):    
    def __init__(self, name, year):
        person.__init__(self,name)
        self.__year=year
        self.__grades=list()

    def add_grade(self,grade):
        self.__grades.append(grade)
        
    def average_grade(self):
        return sum(self.__grades)/len(self.__grades)
    
    def print_grades(self):
        for grade in self.__grades:
            print (grade)
            
    def do_work(self):
        print ("Learning...")

            
class faculty(person):
    def __init__(self, name):
        person.__init__(self,name)
        self.__courses=list()

    def add_courses(self,course):
        self.__courses.append(course)
        
    def print_courses(self):
        for course in self.__courses:
            print (course)

    def do_work(self):
        print ("Teaching...")
```

Now we can safely treat two instances of two different classes in the same way:

```python
a_student=student("Bob",2)
a_teacher=faculty("Mary")

a_student.do_work()
a_teacher.do_work()
```

Note that the parent class, in this case, has some methods that should not be used because they are not implemented. Indeed the parent class should probably never be instanciated either.

```python
a_person=person("John")
a_person.do_work()
```

In object oriented programming, classes that shouldn't be instanciated are known as **abstract** and methods that shouldn't be called are known as **virtual**. Some languages (usually compiled ones) don't allow you to instanciate objects that are abstract or have virtual methods. Python is not as strict and some of the usual constructs of object oriented programming are conventions and not enforced.

It's worth being explicit about why Python's version looks the way it does. Python's object-oriented machinery isn't a separate thing compiled into the language the way it is in C++ or Java — it's mostly convention layered on top of an otherwise ordinary language, which is also why it isn't enforced. Compiled OO languages have real keywords for `virtual` and `abstract` that the compiler checks for you; Python has neither. What we do instead is give the method a body that raises an error the moment it's called — `raise NotImplementedError` — so that a class which *should* have been overridden fails loudly if it wasn't, rather than silently doing nothing. It's the same idea as `virtual`/`abstract`, achieved by convention rather than by the language stopping you.

Often a parent class like `person` is meant purely as an interface, so it may contain almost no working code: just the structure, the methods and data that every child is expected to have. That structure defines the interface, and each child class implements it in its own way.

## Polymorphism

In the scenarios described above, since the parent class `person` defines a `do_work()` method, any derived child class will necessarily have `do_work()` available. For example, I can create a list of people and make them work:


```python
people= [student("Bob",2), faculty("Mary")]

for person in people:
    person.do_work()
```

We note that the code here doesn't have to know anything about `student` or `faculty`. It just has to know about `person`. 

But what if I have a list of people and I want to check if a specific object in the list is an instance of a specific class so I can call methods of that class?

You can simply check if the object is an instance of a specfic class:

```python
isinstance(a_student, student)
```

Therefore calling methods specific to the derived class:

```python
for person in people:
    person.do_work()
    if isinstance(person,student):
        person.add_grade(100.)
    if isinstance(person,faculty):
        person.add_courses("Data 3401")

people[0].print_grades()
people[1].print_courses()
```

## Practice: A Counter Class

Before going on to overloading built-ins, here's a small design exercise worth working through yourself, since it's exactly the kind of thing labs will ask you to do. The task: write a class that acts as a counter incrementable up to a specified maximum, prints an error instead of going over that maximum, and can be reset — and then, as a second step, make sure all of the data it holds is private, with methods to check the current value, check the maximum, and check whether it's already at the maximum.

Rather than write the code out, here's how to think through the design, the way you should think through any small class like this before typing anything:

* **What data does an instance need to remember?** Two things: the maximum it's allowed to reach, and its current value.
* **What does the constructor need?** The maximum has to be set when the counter is created — it isn't something you'd want to guess or hard-code — so `__init__` should take it as an argument. The current value starts at some default (probably 0) and isn't something the caller needs to supply.
* **What does `increment` do?** Bump the current value by one — unless doing so would exceed the maximum, in which case print an error instead of incrementing, the same idiom used elsewhere in this notebook (a printed message, not an exception — we haven't covered those as a way to signal this kind of error).
* **What does making the data private change?** Nothing about the logic above — only that `current` and `maximum` become double-underscore-prefixed and get their own accessor methods (current value, maximum value, and a yes/no "are you at the maximum") rather than being read directly, for exactly the reasons the car example above laid out.

That's the whole design — two data members, a constructor, one method that does the real work, and (for the second version) a handful of accessors. Actually writing it is small, but do write it yourself rather than just reading the description: the point of an exercise like this isn't having the answer, it's confirming you can produce this pattern on your own before it shows up again, bigger, later in the course.
